## Exercise 5 – Differences Between LDA and QDA

We examine the performance differences between **Linear Discriminant Analysis (LDA)** and **Quadratic Discriminant Analysis (QDA)** depending on the Bayes decision boundary and sample size.

### (a) Bayes decision boundary is linear

- **Training set:**  
  LDA assumes linear boundaries, so it is correctly specified. QDA is more flexible (quadratic) and can overfit slightly.
    **LDA may perform slightly better or similar to QDA.**

- **Test set:**  
  Since the true boundary is linear, LDA generalizes better. QDA may overfit small fluctuations in the training set.
    **LDA expected to perform better on the test set.**


### (b) Bayes decision boundary is non-linear

- **Training set:**  
  QDA can model the non-linear (quadratic) boundary exactly. LDA cannot, so QDA achieves lower training error.  
    **QDA better on the training set.**

- **Test set:**  
  If sample size is sufficient, QDA also generalizes better. If the sample size is small, QDA may overfit. LDA will underfit.  
    **QDA usually better on test set if enough data.**


### (c) Effect of increasing sample size \(n\)

- As \(n\) increases:  
  - QDA estimates **more parameters** (covariance for each class).  
  - Larger \(n\) reduces variance in these estimates.  

  **Conclusion:**  
The test accuracy of QDA relative to LDA **improves** as the sample size increases because QDA can better capture non-linear boundaries without overfitting.


### (d) True or False statement

> "Even if the Bayes decision boundary is linear, we will probably achieve a superior test error rate using QDA rather than LDA because QDA is flexible enough to model a linear decision boundary."

**Answer:** False 

**Justification:**

- LDA is correctly specified for a linear boundary.  
- QDA is more flexible, but flexibility comes with estimating more parameters, which may **increase variance**, especially with small sample sizes.  
- Modeling a linear boundary with QDA does not improve performance and may slightly worsen test error due to overfitting.
